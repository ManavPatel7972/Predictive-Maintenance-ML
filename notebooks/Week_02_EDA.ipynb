```python
# check outliers
numeric_col =  df.select_dtypes('number')
numeric_col.columns

for col in numeric_col:
    plt.figure()
    
    sns.boxplot(x=df[col])
```

```python
# for above boxplot 
# outliers = [Rotational speed [rpm],Torque [Nm]] in this columns
```

```python
# remove outliers from both columns

out_clo = ['Rotational speed [rpm]','Torque [Nm]']

for col in out_clo:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3- Q1
     
    min = Q1 - 1.5 * IQR
    max = Q3 + 1.5 * IQR
    
    # (df[col] >= min) & (df[col] <= max)  this give bool dataframe
    
    df = df[(df[col] >= min) & (df[col] <= max)]
```

```python
# after remove it further plot both boxplot

for col in out_clo:
    
    plt.figure()
    
    sns.boxplot(x=df[col])
```
