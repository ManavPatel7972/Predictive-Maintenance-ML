```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
class Logistic:
    def __init__(self,learning_rate=0.01,epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weight = None
        self.bias = None
        
    # sigmoid fun
    def sigmoid(self,z):
        return 1 / (1 + np.exp(-z))

    def fit(self,X,y):
        n_samples,n_features = X.shape
        
        self.weight = np.zeros(n_features)
        self.bias = 0
        
        # gradient decent
        for i in range(self.epochs):
            
            z = np.dot(X,self.weight) + self.bias
            
            y_pred = self.sigmoid(z)
            
            dw = (1/n_samples) * np.dot(X.T,(y_pred - y))
            db = (1/n_samples) * np.sum(y_pred - y)
            
            
            self.weight = self.weight - self.learning_rate * dw
            self.bias = self.bias - self.learning_rate * db
        
        self.coef_ = self.weight.reshape(1, -1)
        self.intercept_ = np.array([self.bias])
    
              
    # def predict(self,X):
    #     lin = np.dot(X, self.weight) + self.bias
    #     prob = self.sigmoid(lin)
        
    #     if(prob >= 0.5): return 1
        
    #     return 0   
    
    
    def predict_proba(self, X):
        linear = np.dot(X, self.weight) + self.bias
        return self.sigmoid(linear)

    def predict(self, X):
        probabilities = self.predict_proba(X)
        return np.where(probabilities >= 0.5, 1, 0)
        
             
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../data/process_classification_data.csv")
```

```python
df = df.iloc[0:,1:]
```

```python
df.head(2)
```

```python
X = df.drop(columns=['Machine failure'],axis=1)
y = df['Machine failure']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=10,stratify=y)
```

```python
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train,y_train)
```

```python
print("SkLearn Coef = ",model.coef_)
print("SkLearn Intercept = ",model.intercept_)
```

```python
model.predict_proba(X_test)
```
